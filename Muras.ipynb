```python
import matplotlib.pyplot as plt

COLS, ROWS = 70, 50
def show_map():

    plt.style.use("dark_background")

    fig, ax = plt.subplots(figsize=(9.0, 6.0))
    ax.set_xlim(0, COLS)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_ylim(0, ROWS)
    ax.set_aspect("equal")
    ax.set_facecolor("#000")
    plt.show()
 
show_map() 
```
