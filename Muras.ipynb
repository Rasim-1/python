```python
import matplotlib.pyplot as plt

COLS, ROWS = 70, 50
def show_map():

    plt.style.use("dark_background")

    fig, ax = plt.subplots(figsize=(9.0, 6.0))
    ax.set_xlim(0, COLS)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_ylim(0, ROWS)
    ax.set_aspect("equal")
    ax.set_facecolor("#000")
    plt.show()
 
show_map() 
```

```python
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches

COLS, ROWS = 70, 50


def show_map(num_squares = 6, size = 3):
    plt.style.use("dark_background")

    fig, ax = plt.subplots(figsize=(9.0, 6.0))
    ax.set_axis_off()
    ax.set_xlim(0, COLS)
    ax.set_ylim(0, ROWS)
    ax.set_aspect("equal")
    ax.set_facecolor("#000")

    for _ in range(num_squares):
        x = random.randint(0, COLS - size)
        y = random.randint(0, ROWS - size)
        ax.add_patch(patches.Rectangle((x, y), size, size, food_color = ("#4ade80")))

    plt.tight_layout()
    plt.show()


show_map()
```
